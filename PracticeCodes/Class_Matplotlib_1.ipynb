```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]
```

```python
plt.scatter(x,y, color='g', marker='o')
plt.show()
```

```python
plt.hist2d(x, y, bins=(len(x), len(y)))
plt.colorbar(label='Count')
plt.show()
```

```python
plt.boxplot(x)
plt.show()
```

```python
plt.bar(x,y, color='#66ffbb')
plt.show()
```

```python
plt.subplot(1,2,1)
plt.bar(x,y, color='g')

plt.subplot(1,2,2)
plt.boxplot(x)
```

```python
plt.subplot(1,2,1)
plt.scatter(x,y, color='g', marker='*')

plt.subplot(1,2,2)
plt.hist2d(x, y, bins=(len(x), len(y)))
plt.colorbar(label='Count')
```

```python
name=['a', 'b','c','d','e']
gpa=[3.5,4,3,2.5,4.2]
```

```python
plt.bar(name, gpa, color='g')
plt.show()
```

```python
plt.plot(name, gpa, color='g', marker='*')
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.subplot(2,2,1)
plt.bar(name, gpa, color='g')

plt.subplot(2,2,2)
plt.scatter(name, gpa, color='b')

plt.subplot(2,2,3)
plt.bar(name, gpa, color='r')

plt.subplot(2,2,4)
plt.boxplot(gpa)
plt.show()
plt.savefig('myFirstGraph.pdf')
#
```
